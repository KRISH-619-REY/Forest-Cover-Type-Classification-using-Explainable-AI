## 10 . Baseline Decision Tree and Random Forest

```python
# ============================================================
# BASELINE DECISION TREE & RANDOM FOREST
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
```

```python
# ============================================================
# BASELINE DECISION TREE
# ============================================================

dt_model = DecisionTreeClassifier(
    random_state=42
)

dt_model.fit(
    X_train_model,
    y_train_final
)

print("Decision Tree training completed.")
```

```python
# ============================================================
# BASELINE RANDOM FOREST
# ============================================================

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1,
    oob_score=True
)

rf_model.fit(
    X_train_model,
    y_train_final
)

print("Random Forest training completed.")
```

```python
# ============================================================
# PREDICTIONS
# ============================================================

# -------------------------
# Decision Tree
# -------------------------

y_pred_dt_train = dt_model.predict(X_train_model)
y_pred_dt_val = dt_model.predict(X_val_model)


# -------------------------
# Random Forest
# -------------------------

y_pred_rf_train = rf_model.predict(X_train_model)
y_pred_rf_val = rf_model.predict(X_val_model)

print("Training and validation predictions generated.")
```

```python
# ============================================================
# EVALUATION FUNCTION
# ============================================================

def evaluate_model(model_name, y_true, y_pred):

    return {
        "Model": model_name,

        "Accuracy": accuracy_score(
            y_true,
            y_pred
        ),

        "Macro Precision": precision_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        ),

        "Macro Recall": recall_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        ),

        "Macro F1": f1_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        ),

        "Weighted F1": f1_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0
        )
    }
```

```python
# ============================================================
# BASELINE VALIDATION RESULTS
# ============================================================

baseline_results = pd.DataFrame([

    evaluate_model(
        "Decision Tree",
        y_val,
        y_pred_dt_val
    ),

    evaluate_model(
        "Random Forest",
        y_val,
        y_pred_rf_val
    )

])

display(
    baseline_results.round(4)
)
```

```python
# ============================================================
# RANDOM FOREST - OOB EVALUATION
# ============================================================

# ------------------------------------------------------------
# OOB Accuracy
# ------------------------------------------------------------

oob_accuracy = rf_model.oob_score_


# ------------------------------------------------------------
# OOB Predictions
# ------------------------------------------------------------

oob_proba = rf_model.oob_decision_function_


# ------------------------------------------------------------
# Convert OOB probabilities to predicted class labels
# ------------------------------------------------------------

oob_pred = rf_model.classes_[
    np.argmax(oob_proba, axis=1)
]


# ------------------------------------------------------------
# OOB Macro F1
# ------------------------------------------------------------

oob_macro_f1 = f1_score(
    y_train_final,
    oob_pred,
    average="macro",
    zero_division=0
)


print("Random Forest OOB Accuracy:", round(oob_accuracy, 4))
print("Random Forest OOB Macro F1:", round(oob_macro_f1, 4))
```

```python
# ============================================================
# TRAINING VS VALIDATION PERFORMANCE
# ============================================================

fit_results = pd.DataFrame([

    {
        "Model": "Decision Tree",

        "Train Accuracy": accuracy_score(
            y_train_final,
            y_pred_dt_train
        ),

        "Validation Accuracy": accuracy_score(
            y_val,
            y_pred_dt_val
        ),

        "Train Macro F1": f1_score(
            y_train_final,
            y_pred_dt_train,
            average="macro",
            zero_division=0
        ),

        "Validation Macro F1": f1_score(
            y_val,
            y_pred_dt_val,
            average="macro",
            zero_division=0
        )
    },

    {
        "Model": "Random Forest",

        "Train Accuracy": accuracy_score(
            y_train_final,
            y_pred_rf_train
        ),

        "Validation Accuracy": accuracy_score(
            y_val,
            y_pred_rf_val
        ),

        "Train Macro F1": f1_score(
            y_train_final,
            y_pred_rf_train,
            average="macro",
            zero_division=0
        ),

        "Validation Macro F1": f1_score(
            y_val,
            y_pred_rf_val,
            average="macro",
            zero_division=0
        )
    }

])


# ------------------------------------------------------------
# Calculate gaps
# ------------------------------------------------------------

fit_results["Accuracy Gap"] = (
    fit_results["Train Accuracy"]
    - fit_results["Validation Accuracy"]
)

fit_results["Macro F1 Gap"] = (
    fit_results["Train Macro F1"]
    - fit_results["Validation Macro F1"]
)


display(
    fit_results[
        [
            "Model",
            "Train Accuracy",
            "Validation Accuracy",
            "Train Macro F1",
            "Validation Macro F1",
            "Accuracy Gap",
            "Macro F1 Gap"
        ]
    ].round(4)
)
```

```python
# ============================================================
# GENERALIZATION GAP SUMMARY
# ============================================================

gap_summary = fit_results[
    [
        "Model",
        "Train Accuracy",
        "Validation Accuracy",
        "Accuracy Gap",
        "Train Macro F1",
        "Validation Macro F1",
        "Macro F1 Gap"
    ]
].copy()

display(
    gap_summary.round(4)
)
```

```python
# ============================================================
# FINAL BASELINE COMPARISON
# ============================================================

final_baseline_results = pd.DataFrame([

    {
        "Model": "Decision Tree",

        "Train Accuracy": accuracy_score(
            y_train_final,
            y_pred_dt_train
        ),

        "Validation Accuracy": accuracy_score(
            y_val,
            y_pred_dt_val
        ),

        "Accuracy Gap": (
            accuracy_score(y_train_final, y_pred_dt_train)
            - accuracy_score(y_val, y_pred_dt_val)
        ),

        "Train Macro F1": f1_score(
            y_train_final,
            y_pred_dt_train,
            average="macro",
            zero_division=0
        ),

        "Validation Macro F1": f1_score(
            y_val,
            y_pred_dt_val,
            average="macro",
            zero_division=0
        ),

        "Macro F1 Gap": (
            f1_score(
                y_train_final,
                y_pred_dt_train,
                average="macro",
                zero_division=0
            )
            -
            f1_score(
                y_val,
                y_pred_dt_val,
                average="macro",
                zero_division=0
            )
        ),

        "OOB Score": np.nan,

        "OOB Macro F1": np.nan
    },

    {
        "Model": "Random Forest",

        "Train Accuracy": accuracy_score(
            y_train_final,
            y_pred_rf_train
        ),

        "Validation Accuracy": accuracy_score(
            y_val,
            y_pred_rf_val
        ),

        "Accuracy Gap": (
            accuracy_score(y_train_final, y_pred_rf_train)
            - accuracy_score(y_val, y_pred_rf_val)
        ),

        "Train Macro F1": f1_score(
            y_train_final,
            y_pred_rf_train,
            average="macro",
            zero_division=0
        ),

        "Validation Macro F1": f1_score(
            y_val,
            y_pred_rf_val,
            average="macro",
            zero_division=0
        ),

        "Macro F1 Gap": (
            f1_score(
                y_train_final,
                y_pred_rf_train,
                average="macro",
                zero_division=0
            )
            -
            f1_score(
                y_val,
                y_pred_rf_val,
                average="macro",
                zero_division=0
            )
        ),

        "OOB Score": oob_accuracy,

        "OOB Macro F1": oob_macro_f1
    }

])

display(
    final_baseline_results.round(4)
)
```

```python
# ============================================================
# DECISION TREE - VALIDATION CLASSIFICATION REPORT
# ============================================================

print("DECISION TREE - VALIDATION")
print("=" * 60)

print(
    classification_report(
        y_val,
        y_pred_dt_val,
        zero_division=0
    )
)
```

## 11. Balanced + Tuned Decision Tree

```python
import numpy as np
import pandas as pd

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.model_selection import GridSearchCV, RandomizedSearchCV

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
```

```python
def evaluate_model(model_name, y_true, y_pred):
    
    return {
        "Model": model_name,
        "Accuracy": accuracy_score(y_true, y_pred),
        "Macro Precision": precision_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        ),
        "Macro Recall": recall_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        ),
        "Macro F1": f1_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        ),
        "Weighted F1": f1_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0
        )
    }
```

```python
dt_base = DecisionTreeClassifier(
    class_weight="balanced",
    random_state=42
)
```

```python
dt_param_grid = {
    "max_depth": [10, 20, 30, None],
    
    "min_samples_split": [2, 10, 20],
    
    "min_samples_leaf": [1, 5, 10],
    
    "max_features": [None, "sqrt"],
    
    "ccp_alpha": [0.0, 0.001]
}
```

```python
dt_grid = GridSearchCV(
    estimator=dt_base,
    param_grid=dt_param_grid,
    scoring="f1_macro",
    cv=3,
    n_jobs=-1,
    verbose=2
)

dt_grid.fit(
    X_train_final,
    y_train_final
)
```

```python
print("BEST DECISION TREE PARAMETERS")
print("=" * 50)

print(dt_grid.best_params_)

print("\nBest CV Macro F1:")
print(dt_grid.best_score_)
```

```python
# ============================================================
# BALANCED + TUNED DECISION TREE - PREDICTIONS
# ============================================================

dt_tuned_balanced = dt_grid.best_estimator_

y_pred_dt_tuned = dt_tuned_balanced.predict(
    X_val_model
)

print("Balanced + Tuned Decision Tree predictions generated.")
```

```python
# ============================================================
# BALANCED + TUNED DECISION TREE - VALIDATION RESULTS
# ============================================================

dt_tuned_balanced_results = pd.DataFrame([
    evaluate_model(
        "Balanced + Tuned Decision Tree",
        y_val,
        y_pred_dt_tuned
    )
])

display(
    dt_tuned_balanced_results.round(4)
)
```

```python
# ============================================================
# XGBOOST — BALANCED HYPERPARAMETER TUNING USING GPU
# ============================================================

from xgboost import XGBClassifier
from sklearn.model_selection import RandomizedSearchCV
from sklearn.utils.class_weight import compute_sample_weight


# ============================================================
# 1. XGBOOST BASE MODEL FOR TUNING
# ============================================================

xgb_model = XGBClassifier(
    objective="multi:softmax",
    num_class=7,
    eval_metric="mlogloss",

    # GPU
    tree_method="hist",
    device="cuda",

    random_state=42,
    n_jobs=-1
)


# ============================================================
# 2. HYPERPARAMETER SEARCH SPACE
# ============================================================

xgb_param_dist = {

    "n_estimators": [100, 150, 200, 250, 300],

    "max_depth": [5, 7, 9, 11],

    "learning_rate": [0.01, 0.05, 0.1, 0.2],

    "min_child_weight": [1, 3, 5, 7],

    "gamma": [0, 0.1, 0.3, 0.5],

    "subsample": [0.7, 0.8, 0.9, 1.0],

    "colsample_bytree": [0.7, 0.8, 0.9, 1.0],

    "reg_alpha": [0, 0.1, 0.5, 1],

    "reg_lambda": [1, 2, 5, 10]
}


# ============================================================
# 3. COMPUTE BALANCED SAMPLE WEIGHTS
# ============================================================

sample_weights_xgb = compute_sample_weight(
    class_weight="balanced",
    y=y_train_xgb
)

print("Balanced sample weights calculated.")

print("Number of training samples:",
      len(sample_weights_xgb))

print("Minimum weight:",
      sample_weights_xgb.min())

print("Maximum weight:",
      sample_weights_xgb.max())


# ============================================================
# 4. RANDOMIZED SEARCH
# ============================================================

xgb_random = RandomizedSearchCV(

    estimator=xgb_model,

    param_distributions=xgb_param_dist,

    n_iter=20,

    scoring="f1_macro",

    cv=3,

    random_state=42,

    n_jobs=1,

    verbose=2,

    return_train_score=False
)


# ============================================================
# 5. FIT BALANCED XGBOOST
# ============================================================

print("\nStarting Balanced XGBoost Hyperparameter Tuning...")
print("GPU: CUDA")
print("Scoring: Macro F1")
print("CV: 3-fold")
print("Candidates: 20")
print()


xgb_random.fit(
    X_train_model,
    y_train_xgb,
    sample_weight=sample_weights_xgb
)


# ============================================================
# 6. BEST PARAMETERS
# ============================================================

print("\n==============================================")
print("BEST BALANCED XGBOOST PARAMETERS")
print("==============================================")

for key, value in xgb_random.best_params_.items():
    print(f"{key}: {value}")


# ============================================================
# 7. BEST CROSS-VALIDATION SCORE
# ============================================================

print("\nBest Cross-Validation Macro F1:")
print(round(xgb_random.best_score_, 4))
```

```python
# ============================================================
# CATBOOST — TUNED + BALANCED
# Forest CoverType Classification
# ============================================================

import numpy as np
import pandas as pd

from catboost import CatBoostClassifier
from sklearn.model_selection import RandomizedSearchCV


# ============================================================
# 1. CATBOOST ESTIMATOR
# ============================================================

cat_model_tuning = CatBoostClassifier(

    # --------------------------------------------------------
    # MULTICLASS CLASSIFICATION
    # --------------------------------------------------------
    loss_function="MultiClass",
    eval_metric="MultiClass",

    # --------------------------------------------------------
    # BALANCING
    # --------------------------------------------------------
    # CatBoost automatically calculates class weights
    # inversely proportional to class frequencies.
    auto_class_weights="Balanced",

    # --------------------------------------------------------
    # GPU
    # --------------------------------------------------------
    task_type="GPU",
    devices="0",

    # --------------------------------------------------------
    # REPRODUCIBILITY
    # --------------------------------------------------------
    random_seed=42,

    # Don't print CatBoost's internal training output
    verbose=0
)


# ============================================================
# 2. HYPERPARAMETER SEARCH SPACE
# ============================================================

cat_param_dist = {

    "iterations": [
        300,
        500,
        700,
        1000
    ],

    "depth": [
        6,
        7,
        8,
        9,
        10
    ],

    "learning_rate": [
        0.03,
        0.05,
        0.08,
        0.10
    ],

    "l2_leaf_reg": [
        1,
        3,
        5,
        7,
        10
    ],

    "random_strength": [
        0,
        0.5,
        1,
        2
    ],

    "bagging_temperature": [
        0,
        0.5,
        1,
        2
    ],

    "border_count": [
        32,
        64,
        128
    ]
}

# ============================================================
# 3. RANDOMIZED SEARCH
# ============================================================

cat_random = RandomizedSearchCV(

    estimator=cat_model_tuning,

    param_distributions=cat_param_dist,

    # Number of random parameter combinations
    n_iter=15,

    # We care about Macro F1 because the dataset is imbalanced
    scoring="f1_macro",

    # 3-fold CV
    cv=3,

    random_state=42,

    # CatBoost itself is using GPU.
    # Keep sklearn parallelism at 1 to avoid
    # multiple GPU jobs fighting for VRAM.
    n_jobs=1,

    verbose=2
)


# ============================================================
# 4. FIT TUNED + BALANCED CATBOOST
# ============================================================

print("=" * 65)
print("STARTING TUNED + BALANCED CATBOOST")
print("=" * 65)

cat_random.fit(
    X_train_model,
    y_train_final
)


# ============================================================
# 5. BEST PARAMETERS
# ============================================================

print("\n" + "=" * 65)
print("BEST CATBOOST PARAMETERS")
print("=" * 65)

print(cat_random.best_params_)


# ============================================================
# 6. BEST CROSS-VALIDATION MACRO F1
# ============================================================

print("\nBest Cross-Validation Macro F1:")
print(round(cat_random.best_score_, 4))


# ============================================================
# 7. FINAL TUNED + BALANCED CATBOOST MODEL
# ============================================================

cat_tuned_balanced = cat_random.best_estimator_

print("\nFinal Tuned + Balanced CatBoost:")
print(cat_tuned_balanced)
```

## 20. Balanced + Tuned AdaBoost

```python
# ============================================================
# 1. IMPORTS
# ============================================================

import pandas as pd
import numpy as np

from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

from sklearn.utils.class_weight import compute_sample_weight

# ============================================================
# 2. BALANCED SAMPLE WEIGHTS
# ============================================================

ada_sample_weights = compute_sample_weight(
    class_weight="balanced",
    y=y_train_final
)

print("Balanced sample weights created.")

# ============================================================
# 3. BASE DECISION TREE
# ============================================================

ada_base_tree = DecisionTreeClassifier(
    random_state=42
)


# ============================================================
# 4. ADABOOST MODEL
# ============================================================

ada_model = AdaBoostClassifier(
    estimator=ada_base_tree,
    random_state=42
)

# ============================================================
# 5. HYPERPARAMETER SEARCH SPACE
# ============================================================

ada_param_dist = {

    "n_estimators": [
        50,
        100,
        150,
        200,
        250,
        300
    ],

    "learning_rate": [
        0.01,
        0.03,
        0.05,
        0.1,
        0.2,
        0.3,
        0.5
    ],

    "estimator__max_depth": [
        1,
        2,
        3
    ],

    "estimator__min_samples_split": [
        2,
        5,
        10,
        20
    ],

    "estimator__min_samples_leaf": [
        1,
        2,
        5,
        10
    ],

    "estimator__max_features": [
        None,
        "sqrt"
    ]
}


# ============================================================
# 6. STRATIFIED CROSS-VALIDATION
# ============================================================

cv_strategy = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)



# ============================================================
# 7. RANDOMIZED SEARCH
# ============================================================

ada_random = RandomizedSearchCV(
    estimator=ada_model,
    param_distributions=ada_param_dist,
    n_iter=12,
    scoring="f1_macro",
    cv=cv_strategy,
    random_state=42,
    n_jobs=-1,
    verbose=2,
    error_score="raise"
)



# ============================================================
# 8. HYPERPARAMETER TUNING
# ============================================================

print("\nStarting AdaBoost hyperparameter tuning...")
print("Balanced training weights are being used.")
print("Scoring metric: Macro F1")
print("CV folds: 3")
print("Randomized candidates: 12")

ada_random.fit(
    X_train_model,
    y_train_final,
    sample_weight=ada_sample_weights
)

```

```python
# ============================================================
# 5. CALCULATE SHAP VALUES
# ============================================================

shap_values = explainer.shap_values(X_shap)

print("SHAP values calculated.")
```

```python
# ============================================================
# 6. CHECK SHAP OUTPUT
# ============================================================

print("Type:", type(shap_values))

if isinstance(shap_values, list):

    print("Number of classes:", len(shap_values))

    for i, values in enumerate(shap_values):
        print(
            f"Class {i + 1}:",
            values.shape
        )

else:

    print(
        "SHAP shape:",
        np.array(shap_values).shape
    )
```

```python
# ============================================================
# 7. GLOBAL SHAP FEATURE IMPORTANCE
# ============================================================

if isinstance(shap_values, list):

    mean_abs_shap = np.mean(
        np.stack([
            np.abs(values)
            for values in shap_values
        ]),
        axis=(0, 1)
    )

else:

    shap_array = np.array(shap_values)

    if shap_array.ndim == 3:

        mean_abs_shap = np.mean(
            np.abs(shap_array),
            axis=(0, 2)
        )

    else:

        mean_abs_shap = np.mean(
            np.abs(shap_array),
            axis=0
        )


shap_importance = pd.DataFrame({
    "Feature": X_shap.columns,
    "Mean |SHAP Value|": mean_abs_shap
})

shap_importance = (
    shap_importance
    .sort_values(
        "Mean |SHAP Value|",
        ascending=False
    )
    .reset_index(drop=True)
)

shap_importance["Rank"] = (
    np.arange(1, len(shap_importance) + 1)
)

shap_importance = shap_importance[
    [
        "Rank",
        "Feature",
        "Mean |SHAP Value|"
    ]
]

display(shap_importance.head(20))
```

```python
# ============================================================
# 8. SHAP GLOBAL SUMMARY PLOT
# ============================================================

shap.summary_plot(
    shap_values,
    X_shap,
    max_display=20
)
```

```python
# ============================================================
# CLASS-SPECIFIC SHAP SUMMARY PLOTS — ALL 7 CLASSES
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import shap

# Number of classes
n_classes = 7

# Class labels
class_labels = [1, 2, 3, 4, 5, 6, 7]


# ============================================================
# GENERATE SHAP PLOT FOR EACH CLASS
# ============================================================

for class_index in range(n_classes):

    print(f"Generating SHAP plot for Class {class_labels[class_index]}...")

    # --------------------------------------------------------
    # Extract SHAP values for the current class
    # --------------------------------------------------------

    if isinstance(shap_values, list):

        # Older / some SHAP versions:
        # shap_values[class_index]
        shap_class = shap_values[class_index]

    else:

        shap_array = np.array(shap_values)

        # Shape:
        # (samples, features, classes)
        if shap_array.ndim == 3:

            shap_class = shap_array[:, :, class_index]

        # Binary / single-output case
        else:

            shap_class = shap_array


    # --------------------------------------------------------
    # Create separate figure
    # --------------------------------------------------------

    plt.figure(figsize=(10, 7))

    shap.summary_plot(
        shap_class,
        X_shap,
        max_display=15,
        show=False
    )

    plt.title(
        f"SHAP Feature Importance — Class {class_labels[class_index]}",
        fontsize=14
    )

    plt.tight_layout()

    plt.show()
```
